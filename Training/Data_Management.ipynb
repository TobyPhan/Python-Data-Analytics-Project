```python
#Date Cleaning. #Wants non values filled in with some sort of median value. Also remove any duplicate job entries

import pandas as pd                        #Dataset
from datasets import load_dataset
import matplotlib.pyplot as plt  

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
#Fill in values, replacing null

median_salary_year = df['salary_year_avg'].median()
median_salary_hour = df['salary_hour_avg'].median()

df_filled = df
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)   #fillna
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
df_filled.loc[:10,'salary_hour_avg':'salary_year_avg']


df_unique = df_filled                              

df_unique = df_unique.drop_duplicates(subset = ['job_title','company_name']) #filter based on job title and company name (unique)
len(df_filled)-len(df_unique)
```
