```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_salary = df[pd.notna(df['salary_year_avg'])] #filter out na of salary year average and provide the salary year average column

#We want to calcualte the projected salary for next year. Assume 3% inflation
#func. function to apply to each column or row. Apply function

def projected_salary(salary):
    return salary * 1.03
df_salary['salary_year_inflated'] = df_salary['salary_year_avg'].apply(projected_salary)
df_salary[['salary_year_avg','salary_year_inflated']]

#For providing the apply function, we can use an anonymous function

df_salary['salary_year_inflated']= df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)
df_salary[['salary_year_avg','salary_year_inflated']]
   
```

```python
# df_salary['salary_year_inflated']= df_salary['salary_year_avg'].apply(lambda salary: salary * 1.03)
# df_salary[['salary_year_avg','salary_year_inflated']]

#Convert from string to a list

type(df['job_skills'][1])

list(df['job_skills'][1])   #Could do this, but it will return to a jumble mess to convert every letter in a list

#Instead we use the ast,literal_eval (abstract syntax trees)        Provide a node or string 

import ast 

ast.literal_eval(df['job_skills'][1])  
#now its a list. Now lets go back and clean up the column using the apply function

def clean_list(skill_list):
    if pd.notna(skill_list):
        return ast.literal_eval(skill_list)                   #Wont fully work if the column has NA, so we will use if statement

df['job_skills'] = df['job_skills'].apply(clean_list)          
type(df['job_skills'][1])
```
