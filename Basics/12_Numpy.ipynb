```python
import random

salary_list = [random.randint(50000,100000)  for i in range(1_000_000)]  #Random List of over 1,000,000 values
```

```python
import statistics

%%timeit  #timing the operatiion. Run multiple time and provide us the average time of running the code

statistics.mean(salary_list)                 #Takes only 300 ms
```

```python
import numpy as np

%%timeit

np.mean(salary_list)                    #Takes only 50 ms. Numpy runs faster than the statistics
```
