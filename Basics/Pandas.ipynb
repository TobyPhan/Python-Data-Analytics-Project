```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
df[(df.job_title_short == 'Data Analyst') & (df.salary_year_avg.notna())]
```

```python

df.iloc[2]  #gets the 2nd row

df['job_country'].unique()


filtered_df = df[df['job_schedule_type'] == 'Full-time']
filtered_df.head(3)

not_na_values_df = df[df['job_skills'].notna()]
not_na_values_df
```

```python
df['job_posted_date']=pd.to_datetime(df.job_posted_date)   #conversion to datetime
```

```python
df.job_posted_date.dt.month           #takes out month
df['job_posted_month'] = df.job_posted_date.dt.month           #takes out month
```

```python
df.sort_values('job_posted_date')

df.sort_values('job_posted_date', inplace=True)              #inplace can pernamently remove or change

df.drop(labels = 'salary_hour_avg', axis = 1, inplace = True)      #dropping the column

df.dropna(subset=['salary_year_avg'], inplace = True)
```

```python
df.sort_values('salary_year_avg', ascending=False,inplace=True)
df.head(10)

df['job_posted_date']= pd.to_datetime(df['job_posted_date'])
df['job_posted_year'] = df.job_posted_date.dt.year

df.dropna(subset=['salary_year_avg'], inplace=True)

```

```python
df.describe()
df.info()
df.count()
df['salary_year_avg'].median()
df['salary_year_avg'].min()

min_salary = df['salary_year_avg'].idxmin()  #provide the index value/row number of the minimum salary
df.iloc[min_salary]


df['job_title_short'].unique()
df['job_title_short'].value_counts()

df.groupby('job_title_short')['salary_year_avg'].min()    #First part groups, second part specify what columns we want to do the aggregation of a stat on

df.groupby(['job_title_short','job_country'])['salary_year_avg'].min()     #then groups by country too. <Make sure the things that you are grouping is wihtin a list

df.groupby('job_title_short')[['salary_year_avg','salary_hour_avg']].min()  # Then add in more columns in the 2nd part (Dont forget to put those into lists too)

df.groupby('job_title_short')[['salary_year_avg','salary_hour_avg']].agg(['min','max','median'])  # Multuiple operations/aggregation methods on it. (Dont forget to put those into lists too)
```

```python
print(df['salary_year_avg'].mean())
print(df['salary_year_avg'].median())

df.groupby('job_title_short')['salary_hour_avg'].mean()    #General way to groupby

min_hourly_index = df['salary_hour_avg'].idxmin()
df.iloc[min_hourly_index][['job_title_short','salary_hour_avg']]     #gets index and then returns specific coloumns


df.groupby('job_country').size()   #Just returns the count

```

```python
import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date']=pd.to_datetime(df.job_posted_date)


df['job_country'].value_counts().head(20)

df['job_country'].isin(['Brazil']).any()   #see if Brazil is in the dataset

us_jobs = df[df['job_country'] == "United States"]            #create a new dataframe that filers to a country of choice

us_jobs = us_jobs[us_jobs['salary_year_avg'].notna()]            #More filters to include data without NAN

us_jobs

us_jobs.groupby('job_title_short').count()  #Group by job title and gets count of every. Min wont work

us_jobs.groupby('job_title_short')['salary_year_avg'].min()

us_jobs.groupby('job_title_short')['salary_year_avg'].agg('median')

us_jobs.groupby('job_title_short')['salary_year_avg'].agg(['min','median','max','count'])   #For more aggregations, put them into a list

us_jobs.groupby('job_title_short')['salary_year_avg'].agg(['median','min','max','count']).sort_values(by = 'median', ascending = False)   #Sort by median, descending

```
