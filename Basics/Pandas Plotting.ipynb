```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date']=pd.to_datetime(df.job_posted_date)
```

```python
job_counts = df.job_title_short.value_counts()
plt.bar(job_counts.index, job_counts)         #Annoying to remember this with the index value, etc

                                                     #For pandas we have plotting. plot(data, x ,y , kind, etc)
job_counts.plot(kind='bar')
plt.title('Postings by Job Title')
plt.xlabel('')
plt.ylabel('Count of Job Postings')
plt.xticks(rotation = 45, ha = 'right')
plt.show()                                      #Formatting will the same as matplot.




df[['job_posted_date','salary_year_avg']].dropna(subset=['salary_year_avg'])
df.plot(x = 'job_posted_date', y = 'salary_year_avg', kind = 'line')          #Doing this, but on dataframe. We need to provide x and y parameters
```

```python
platforms = df['job_via'].value_counts().head(5)
platforms.plot(kind = 'bar')
plt.show

top_5 = df['job_location'].value_counts().head(5).sort_values(ascending = True)
top_5.plot(kind = 'barh')
plt.xlabel('Count of Jobs')
plt.ylabel('Job Location')
plt.title('Number of Job Postings by Job Location')
plt.show

monthly_job_counts = df['job_posted_date'].dt.to_period('M').value_counts().sort_index()
monthly_job_counts.plot(kind = 'line')
plt.xlabel('Month')
plt.ylabel('Number of Job Postings')
plt.title('Trend of Job Postings Over Time')
plt.show
```
